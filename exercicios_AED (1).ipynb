```python
#Ex 1 

def patos_e_coelhos(cabecas, pe):
    coelhos = (pe - 2 * cabecas) / 2
    patos = cabecas - coelhos
    return patos, coelhos

cabecas = int(input("Total de cabeças: "))
pe = int(input("Total de pés: "))

patos, coelhos = patos_e_coelhos(cabecas, pe)

print("Total de patos:", patos)
print("Total de coelhos:", coelhos)
```

```python
#Ex 2

ra = input("Digite o RA: ")

soma = 0
multiplicacao = 1

for digito in ra:
    soma = soma + int(digito)
    multiplicacao = multiplicacao * int(digito)

print("Soma =", soma)
print("Multiplicação =", multiplicacao)
```

```python
#Ex 3

def reverso(numero):
    numero_string = str(numero)
    revertido = numero_string[::-1]
    return int(revertido)

numero = int(input("Digite um número: "))
print("Reverso:", reverso(numero))
```
